# 游戏规则介绍

- 战场为 17x17 网格地图，**两名玩家各控制两个角色**。角色出生点位于<mark>主对角线或副对角线</mark>的两端。

- 地图上存在障碍物，炸弹以及NPC
   - 障碍物所在的网格不可到达
   - NPC 与玩家角色一样行动，参与金币争夺
   - 角色移动到炸弹网格，损失当前持有金币的 X%（向上取整,当前为10%），炸弹消失，角色可继续行动
   - 每 Y 轮（当前为20）刷新一次炸弹
- 地图信息
   - 全局信息不对玩家公开，每个角色默认视野为以自身为中心的 5x5 网格，视野范围内的网格信息对玩家可见
   - 玩家可以花费金币扩展视野范围
   - 每D轮（当前为5）发布全局各区域的快照信息
- NPC
  - 开局时所有 NPC 出生在地图最中央，总数为 A（公测期间为7）
  - 每个 NPC 每轮最多移动 B （当前为3）步
  - NPC 之间、NPC 与玩家角色之间可重叠
  - NPC 参与金币争夺，但不计入选手胜负判定
- 金币生成：
  - 中心 9x9 区域每回合随机生成一定数量金币
  - 每若干轮在 9x9 区域外随机生成一定数量金币
- 移动与拾取规则：
  - 每名玩家的两个角色初始金币均为 0
  - 每轮两个角色共分配 S （当前为6）次移动机会，可任意分配。非法移动（越界、碰障碍物、碰其他玩家角色）不执行，后续移动照常进行
  - 玩家角色进入有超过 2 个 NPC 的格子时，因拥挤踩踏损失当前 N%（当前为5%） 金币（向上取整）
  - 每回合按决策耗时从小到大依次执行移动。同一网格的金币/炸弹由先执行者获取/触发。NPC 行动顺序位于两名玩家之间
  - 拾取规则：角色或 NPC 从其他网格移动到有金币的网格，获得该格 C%（当前为65%）金币，向上取整

- 获胜规则
  - 每场比赛固定500轮，结束时金币总量最多的玩家获胜。金币总数相等时，P90延迟更低的玩家获胜(百分之九十`moveDecision`响应时间更快，即比较第百分之九十分位数)。
